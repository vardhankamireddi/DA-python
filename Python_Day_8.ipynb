```python
#perfect number or not
n=int(input())
sum =0
for i in range (1,n):
  if n % i == 0:
    print(i)
    sum = sum + i
if n == sum:
  print("is perfect number")
else:
  print("Not a perfect number")
```

```python
#prime numbers
num=int(input())
if num>=2:
  for i in range(2,num):
    if num % i == 0:
      print("Not a prime number")
      break
  else:
    print("it is prime number")
```

```python
# factors of number
num = int(input())
for i in range(1,num+1):
    if num % i == 0:
        print(i, end =" ")
```
