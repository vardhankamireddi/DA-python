Set Methods


1.   Add()
2.   Update()
3.   Remove()
4.   Discard()
5.   pop()
6.   Clear()
7.   Copy()


```python
# Set methods add()
s={1,3,5,8,5,3,67,9}
s.add(10)
print(s)
```

```python
#set methods update()
b={10,20,30,40}
b.update([50,"vardhan"])
print(b)
```

```python
#set methods remove()
c={10,20,50,30,70}
c.remove(20)
print(c)
```

```python
#set methods discard()
d={10,20,4,5,9}
d.discard(9)
print(d)
```

```python
#set methods pop()
d={20,40,50,30,10}
d.pop()
print(d)
```

```python
#set methods clear()
a={10,40,30,60}
a.clear()
print(a)
```

```python
#set methods Copy()
a={10,20,50,70,80,90}
b=a.copy()
print(b)
```

Mathematical OPerations on sets


*   Union(|) removes duplicates and prints the set
*   Intersection(&) prints Only the common
*   difference(-) removes common and prints first set elements
*   Symetric difference(^) removes common and prints all elemtns from two sets



```python
 #union(|)
a={10,20,40,30,50}
b={20,50,30,60,90}
print(a|b)
```

```python
#Intersection(&)
print(a&b)
```

```python
#difference(-)
print(a-b)
print(b-a)
```

```python
#symmetric Difference(^)
print(a^b)
```
