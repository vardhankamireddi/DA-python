```python
# count uppercase and lowercase and digit
a=input()
u = 0
l = 0
d = 0
for i in a:
  if i.isupper():
    u = u+1
  elif i.islower():
    l = l+1
  elif i.isdigit():
    d = d+1
  else:
    pass
print("Lower count:",l)
print("Upper count:",u)
print("Digit count:",d)
```
