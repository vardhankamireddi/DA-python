```python
student = {"name":"vardhan","age":"22","marks":"99"}
print(student)
print(student["name"])
print(student["age"])
print(student["marks"])
print(type(student))
```

```python
#adding elements to the dictionary
student = {"name":"vardhan","age":"22","marks":"99"}
student["course"] = "Data anaytics"
student["marks"] = 100
print(student)
```

```python
#updating dictionary
student = {"name":"vardhan","age":"22","marks":"99"}
student["age"] = 23
student["marks"] = 90
print(student)
```

```python
#create a dictionary of employees details with five details
employee = {"name":"vardhan","age":32,"salary":45000,"gender":"Male","role":"Software Employee"}
employee["salary"] = 120000

print(employee)
```

```python
#add employee blood group
employee = {"name":"vardhan","age":32,"salary":45000,"gender":"Male","role":"Software Employee"}
employee["salary"] = 120000
employee["Blood Group"] ="o+"
print(employee)
```

```python
#removing employee dict elements
employee = {"name":"vardhan","age":32,"salary":45000,"gender":"Male","role":"Software Employee"}
employee["salary"] = 120000
employee["Blood Group"] ="o+"
employee.pop("age")
employee.pop("salary")
employee.pop("Blood Group")
employee.popitem()
print(employee)
```

```python
#for getting only keys
employee = {"name":"vardhan","age":32,"salary":45000,"gender":"Male","role":"Software Employee"}
employee["salary"] = 120000
employee["Blood Group"] ="o+"
for i in employee.keys():
  print(i)
```

```python
# for getting both keys and values of dict
for i,j in employee.items():
  print(f"{i} : {j}")
```

```python
# for getting values from dict
for i in employee.values():
  print(i)
```

```python
dis=int(input())
if dis<=2:
  fare=50
elif dis <=10:
  fare=50+(dis-2) * 15
else:
  fare=170+(dis-10) * 10
if dis > 20:
  total_fare=fare*0.95
else:
  total_fare=fare
print("Total fare:",total_fare)
```
