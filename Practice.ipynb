# Dictionary



1.   write a python program to add a new key "city" with the value "hyderabad" to a given dictionary
2.   write a python program to update the salary of an employee to 50000

3. write a python program to remove the key "age" from the given dictionary

4.  write a python program to display all the keys and values of a given dictionary

5.  write a python program to check whether the key "department" is present in a given dictionary



```python

# write a python program to add a new key "city" with the value "hyderabad" to a given dictionary
a={"name":"vardhan","age":"22"}
a["city"]="hyderabad"
print(a)
```

```python
# write a python program to update the salary of an employee to 50000
employee={"name":"vardhan ","age":"22","salary":"30000"}
employee["salary"]="50000"
print(employee)
```

```python
 #write a python program to remove the key "age" from the given dictionary
employee={"name":"vardhan","age":"22","salary":"30000"}
employee.pop("age")
print(employee)
```

```python
 #write a python program to display all the keys and values of a given dictionary
employee={"name":"vardhan","age":"22","salary":"30000"}
for i,j in employee.items():
  print(f'{i}:{j}')
```
