```python
# leap year or not
x= int(input())
if (x%4==0):
  print(f"{x} leap year")
elif (x%400==0 and x%100 !=0):
  print(f"{x} leap year")
else:
  print(f"{x} not a leap year")
```

```python
maths, science, english = map(int, input().split())
total = (maths + science + english)
print(f"Total: {total}")
avg = total / 3
print(f"Average: {avg}")

if maths < 35 or science < 35 or english < 35:
    print("Fail")
elif avg >= 90 and avg<=100:
    print("Grade A")
elif avg >= 75 and avg <= 89:
    print("Grade B")
elif avg >= 60 and avg <= 74:
    print("Grade C")
elif avg >= 40 and avg <= 59:
    print("Grade D")
else:
    print("Grade F")
```
