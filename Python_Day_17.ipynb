```python
# create a tuple with five fruits by using tuple method

a = tuple(input().split())
print(a)
```

```python
# access the element of last fruit

print(a[-1])
```

```python
# print the fruits in reverse order

print(a[::-1])
```

```python
# converting list to tuple

l= list(a)
print(type(l))
t=tuple(l)
print(type(t))
```

```python
# create a tuple containing 10 numbers and print the first,last and middle elements

t = tuple(input().split())
print(t)
print(t[0])
print(t[4])
print(t[-1])
```

```python
# write a program to find the length of a tuple and count how many times the number 20 occurs in it
c = tuple(input().split())
print(c)
print(type(c))
print(len(c))
print(c.count("20"))
```

```python
# given a tuple of numbers convert it into a list replace the second element with 100, and convert it back into a tuple

t = (10,20,30,40,50,60,70)
print(type(t))
l=list(t)
print(type(l))
l[1]=100
print(l)
a=tuple(l)
print(type(a))
```
