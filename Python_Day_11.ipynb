```python
pin=int(input())
while True:
  if pin < 999 or pin > 99999:
    print ("choose 4 or 6 Digit")
    break
  password = int(input())
  if pin == password:
    print("Login OK")
    break
  else:
    print("Login unsucessfull try Again")
```

```python
pin = input("register your pin = ")
while True:
  if len(pin) == 4 or len(pin) == 6:
    print ("ALL OK")
    break
  else:
    print("enter 4 or 6 digit")
while True:
  lock = input("enter lock = ")
  if pin == lock:
    print("lock ok")
    break
  else:
    print("Not ok")
```

```python
a=int(input())
b=int(input())
i=1
while i <= 5:
    ope = input()
    if ope in "!,@,#,&,(,),{,},[,]":
      print("Not Accecpted")
      break
    elif ope == "*":
      print(a*b)
    elif ope == "+":
      print(a+b)
    elif ope == "-":
      print(a-b)
    elif ope == "**":
      print(a**b)
    elif ope == "%":
      print(a%b)
    elif ope == "/":
      print(a/b)
    elif ope == "//":
      print(a//b)
    else:
      print("Nothing")
    i=i+1
```
